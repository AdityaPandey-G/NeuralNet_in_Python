```python
import Neuron as N
```

```python

class Linear:
    def __init__(self , n_in , n_out):
        self.neurons = [N.Perceptron(n_in) for _ in range(n_out)]

    def __call__(self , x):
        return [n(x) for n in self.neurons]
```

```python
x = [1 , 2 , 3 , 4 , 5]
a = Linear(len(x) , 4)
a(x)
```
